Импортируем библиотеку pandas для работы с табличными данными. Загружаем два файла: нормализованную базу клиентов (xlsx) и файл транзакций (csv).

```python
import pandas as pd

df_norm = pd.read_excel("/Users/valentinayerokhina/Desktop/MyPythonProjects/Capstone/Data/Raw/clients_normalized.xlsx")
df_csv = pd.read_csv("/Users/valentinayerokhina/Desktop/MyPythonProjects/Capstone/Data/Raw/Новые клиенты, капстоун - Лист1.csv", header=None)
print(df_csv.head(3))

print(df_norm.head(3))
```

В файле транзакций отсутствуют заголовки колонок. Присваиваем названия вручную, чтобы в дальнейшем обращаться к колонкам по понятным именам.

```python
df_csv.columns = ["Дата", "Номер_чека", "Категория", "ФИО", "Сумма_без_скидки", "Сумма_со_скидкой"]
print(df_csv.head(3))
```

В таблице df_norm ФИО разбито на три отдельных колонки. Создаём новую колонку ФИО, склеивая их через пробел. Заменяем пустые значения (NaN) на пустую строку через fillna(), убираем лишние пробелы по краям через strip(). Это нужно для JOIN по ФИО — единственной общей колонке в обеих таблицах.

```python
df_norm["ФИО"] = (df_norm["Фамилия"].fillna("")+" "+df_norm["Имя"].fillna("")+" "+df_norm["Отчество"].fillna("")).str.strip()
```

```python
#проверка
print(df_norm.head())
```

```python
#проверка на дубли
df_norm[df_norm["ФИО"].duplicated(keep=False)].sort_values("ФИО").head(10)
```

```python
#удаление дублей
df_norm_dedup = df_norm.drop_duplicates(subset="ФИО", keep="first")
print("Было:", len(df_norm))
print("Стало:", len(df_norm_dedup))
```

Импортируем библиотеку hashlib для хеширования. Создаём уникальный customer_id для каждого клиента: берём ФИО + телефон, кодируем в байты, применяем алгоритм MD5 и получаем 32-символьный уникальный код. Это нужно для анонимизации данных и идентификации клиента в дальнейшем анализе.

```python
import hashlib

df_norm_dedup["customer_id"] = df_norm_dedup.apply(
    lambda row: hashlib.md5(
        (str(row["ФИО"]) + str(row["Телефон_норм"])).encode()
    ).hexdigest(), axis=1
)

print(df_norm_dedup[["ФИО", "Телефон_норм", "customer_id"]].head(3))
```

Создаём итоговую таблицу df_result: присоединяем customer_id из нормализованной таблицы к транзакциям по колонке ФИО. Используем LEFT JOIN — все транзакции остаются, даже если клиент не нашёлся в нормализованной таблице (в таком случае customer_id будет NaN)

```python
df_result = df_csv.merge(df_norm_dedup[["ФИО", "customer_id", 'Возраст']], on="ФИО", how="left")

print(df_result.head(5))
print("Всего строк:", len(df_result))
print("Без customer_id:", df_result["customer_id"].isna().sum())
```

**Анализ записей без customer_id**

После JOIN 3500 транзакций (~5%) остались без customer_id. 

**Причины:**
- Казахские буквы в ФИО (ғ, ә, і) не совпадают с русскими в нормализованной таблице
- Разный регистр букв в отчестве
- Неполные имена (например: "Айгелик -")
- Клиенты чьи визиты были раньше чем создана нормализованная таблица

**Решение:** оставляем как есть — 8% от выборки не влияет 
существенно на результаты анализа. Это задокументированное 
ограничение данных.

```python
no_id = df_result[df_result["customer_id"].isna()]
print(no_id["ФИО"].head(20))
```

Удаляем технические дубли: записи с Категорией 'Проверка зрения' и нулевой суммой — это дубли созданные оптометристом. Реальная стоимость уже внесена консультантом в категории 'Продажа' в тот же день.

```python
df_result["Дата_только"] = df_result["Дата"].str.split(" ").str[0]
```

```python
# находим пары: customer_id + дата где есть и проверка зрения с 0 и продажа
проверки = df_result[
    (df_result["Категория"] == "Проверка зрения") & 
    (df_result["Сумма_без_скидки"] == "0")
][["customer_id", "Дата_только"]]

# помечаем эти пары
df_result["является_парой"] = df_result.set_index(
    ["customer_id", "Дата_только"]
).index.isin(проверки.set_index(["customer_id", "Дата_только"]).index)

print(df_result["является_парой"].sum())
```

```python
df_result.loc[
    (df_result["является_парой"] == True) & 
    (df_result["Категория"] == "Продажа"), 
    "Категория"
] = "Проверка зрения"

print(df_result["Категория"].value_counts())
```

```python
df_result2 = df_result[~(
    (df_result["Категория"] == "Проверка зрения") & 
    (df_result["Сумма_без_скидки"] == "0")
)]

print(df_result2["Категория"].value_counts())
```

```python
print("Строк до:", len(df_result))
print("Строк после очистки:", len(df_result2))
```

Удаляем служебные колонки Дата_только и является_парой — они создавались временно для переименования категорий и не нужны для дальнейшего анализа. 

```python
df_result2 = df_result2.drop(columns=["Дата_только", "является_парой"])
```

```python
#Анонимизируем данные и удаляем клиентов без ID
# Удаляем без customer_id
df_result2 = df_result2[df_result2['customer_id'].notna()]

# Удаляем ФИО
df_result2 = df_result2.drop(columns=['ФИО'])

# Сохраняем
df_result2.to_csv('../data/cleaned_transactions1.csv', index=False)
```
